```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")
df = df[["gender","survived"]]
df.dropna(inplace=True)
total = len(df)
print(total)
```

```python
total_female = len(df[df["gender"]=="female"])
print("Total female : ", total_female)
total_male = len(df[df["gender"]=="male"])
print("Total male : ", total_male)

p_female = total_female/total
p_male = total_male/total
                  
```

```python
# marginal_probability of survival

total_survived = len(df[df["survived"]==1])
print("Total Survived :",total_survived)

total_do_not_survived = len(df[df["survived"]==0])
print("Total Did Not Survive :",total_do_not_survived)

p_survived = total_survived/total
p_not_survived = total_do_not_survived/total
print("Probability of Surviving :",p_survived)
print("Probability of Not Surviving :",p_not_survived)
```

```python
print("\nMarginal Probabilities :")
print(f"P(gender == female):{p_female:.3f}")
print(f"P(gender == male):{p_male:.3f}")
print(f"P(survived == 1):{p_survived:.3f}")
print(f"P(survived == 0):{p_not_survived:.3f}")
```

```python
#joint probability
p_female_survived = len(df[(df["gender"]=="female") & (df["survived"]==1)])/total
p_female_not_survived = len(df[(df["gender"]=="female") & (df["survived"]==0)])/total

p_male_survived = len(df[(df["gender"]=="male") & (df["survived"]==1)])/total
p_male_not_survived = len(df[(df["gender"]==" male") & (df["survived"]==0)])/total
```

```python
print("\nJoint Probabilities :")
print(f"P(female, survived == 1):{p_female_survived:.3f}")
print(f"P(female, survived == 0):{p_female_not_survived:.3f}")
print(f"P(male, survived == 1):{p_male_survived:.3f}")
print(f"P(male, survived == 0):{p_male_not_survived:.3f}")
```

```python
# conditional probability of survival
p_survived_given_female = p_female_survived/p_female
p_survived_given_male = p_male_survived/p_male  
p_not_survived_given_female = p_female_not_survived/p_female
p_not_survived_given_male = p_male_not_survived/p_male
print("\nConditional Probabilities :")
print(f"P(survived == 1 | gender == female):{p_survived_given_female:.3f}")
print(f"P(survived == 1 | gender == male):{p_survived_given_male:.3f}")
print(f"P(survived == 0 | gender == female):{p_not_survived_given_female:.3f}")
print(f"P(survived == 0 | gender == male):{p_not_survived_given_male:.3f}")
```

```python
sachin = pd.read_csv("sachin_tendulkar.csv")
sachin
```

```python
zoro = len(sachin[sachin["Runs"]==0])
total_innings = len(sachin)
prob_zoro = zoro / total_innings

shrilanka = len(sachin[sachin["Versus"]=="Sri Lanka"])
prob_shrilanka = shrilanka / total_innings

more_than_50 = len(sachin[sachin["Runs"]>=50])
prob_more_than_50 = more_than_50 / total_innings

bolls_faced_more_than_30 = len(sachin[sachin["Balls_faced"]>30])
prob_bolls_faced_more_than_30 = bolls_faced_more_than_30 / total_innings

print("Marginal Probabilities :")
print(f"Probability of getting zoro : {prob_zoro:.3f}")
print(f"Probability of playing against Sri Lanka : {prob_shrilanka:.3f}")
print(f"Probability of scoring more than 50 runs : {prob_more_than_50:.3f}")
print(f"Probability of facing more than 30 balls : {prob_bolls_faced_more_than_30:.3f}")
```

```python
Shri_Runs_50 = len(sachin[(sachin["Runs"]>=50) & (sachin["Versus"]=="Sri Lanka")])/total_innings
runs_0_dismissed = len(sachin[(sachin["Runs"] == 0) & (sachin["mode_of_dismissal"].str.startswith("c"))]) / total_innings
runs_30_balls_30 = len(sachin[(sachin["Runs"]>30) & (sachin["Balls_faced"]>30)])/total_innings
Pak_dismissed_lbw = len(sachin[(sachin["Versus"] == "Pakistan") & (sachin["mode_of_dismissal"].str.startswith("lbw"))]) / total_innings

#Joint Probability of scoring more than 50 runs and playing against Sri Lanka
print("\nJoint Probability :")
print(f"P(scoring more than 50 runs and playing against Sri Lanka):{Shri_Runs_50:.3f}")
print(f"P(scoring 0 runs and being dismissed by catch):{runs_0_dismissed:.3f}")
print(f"P(scoring more than 30 runs and facing more than 30 balls):{runs_30_balls_30:.3f}")
print(f"P(playing against Pakistan and being dismissed by lbw):{Pak_dismissed_lbw:.3f}")
```

```python
print("\nConditional Probabilities :")

shri = len(sachin[(sachin["Versus"]=="Sri Lanka")])
least_50_runs_given_shri = len(sachin[(sachin["Runs"]>=50) & (sachin["Versus"]=="Sri Lanka")])
p_least_50_runs_given_shri = least_50_runs_given_shri / shri

morethan_30_balls = len(sachin[(sachin["Balls_faced"]>30)])
morethan_30_balls_morethan_30_runs = len(sachin[(sachin["Runs"]>30) & (sachin["Balls_faced"]>30)])
p_morethan_30_balls_morethan_30_runs = morethan_30_balls_morethan_30_runs / morethan_30_balls

runs_0 = len(sachin[sachin["Runs"]==0])
caught = len(sachin[(sachin["mode_of_dismissal"].str.startswith("c")) & (sachin["Runs"]==0)])
p_runs_0_caught = caught / runs_0

england = len(sachin[(sachin["Versus"]=="England")])
least_25_balls_eng = len(sachin[(sachin["Balls_faced"]>=25) & (sachin["Versus"]=="England")])
p_least_30_runs_given_eng = least_25_balls_eng / england

print(f"P(scoring more than 50 runs | playing against Sri Lanka):{p_least_50_runs_given_shri:.3f}")
print(f"P(scoring more than 30 runs | facing more than 30 balls):{p_morethan_30_balls_morethan_30_runs:.3f}")
print(f"P(scoring 0 runs | being dismissed by catch):{p_runs_0_caught:.3f}")
print(f"P(scoring more than 25 balls | playing against England):{p_least_30_runs_given_eng:.3f}")
```
