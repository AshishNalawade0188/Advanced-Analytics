```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")
df = df[["gender","survived"]]
df.dropna(inplace=True)
total = len(df)
print(total)
```

```python
total_female = len(df[df["gender"]=="female"])
print("Total female : ", total_female)
total_male = len(df[df["gender"]=="male"])
print("Total male : ", total_male)

p_female = total_female/total
p_male = total_male/total
                  
```

```python
# marginal_probability of survival

total_survived = len(df[df["survived"]==1])
print("Total Survived :",total_survived)

total_do_not_survived = len(df[df["survived"]==0])
print("Total Did Not Survive :",total_do_not_survived)

p_survived = total_survived/total
p_not_survived = total_do_not_survived/total
print("Probability of Surviving :",p_survived)
print("Probability of Not Surviving :",p_not_survived)
```

```python
print("\nMarginal Probabilities :")
print(f"P(gender == female):{p_female:.3f}")
print(f"P(gender == male):{p_male:.3f}")
print(f"P(survived == 1):{p_survived:.3f}")
print(f"P(survived == 0):{p_not_survived:.3f}")
```

```python
#joint probability
p_female_survived = len(df[(df["gender"]=="female") & (df["survived"]==1)])/total
p_female_not_survived = len(df[(df["gender"]=="female") & (df["survived"]==0)])/total

p_male_survived = len(df[(df["gender"]=="male") & (df["survived"]==1)])/total
p_male_not_survived = len(df[(df["gender"]==" male") & (df["survived"]==0)])/total
```

```python
print("\nJoint Probabilities :")
print(f"P(female, survived == 1):{p_female_survived:.3f}")
print(f"P(female, survived == 0):{p_female_not_survived:.3f}")
print(f"P(male, survived == 1):{p_male_survived:.3f}")
print(f"P(male, survived == 0):{p_male_not_survived:.3f}")
```

```python
# conditional probability of survival
p_survived_given_female = p_female_survived/p_female
p_survived_given_male = p_male_survived/p_male  
p_not_survived_given_female = p_female_not_survived/p_female
p_not_survived_given_male = p_male_not_survived/p_male
print("\nConditional Probabilities :")
print(f"P(survived == 1 | gender == female):{p_survived_given_female:.3f}")
print(f"P(survived == 1 | gender == male):{p_survived_given_male:.3f}")
print(f"P(survived == 0 | gender == female):{p_not_survived_given_female:.3f}")
print(f"P(survived == 0 | gender == male):{p_not_survived_given_male:.3f}")
```

```python
print("\nConditional Probabilities :")

shri = len(sachin[(sachin["Versus"]=="Sri Lanka")])
least_50_runs_given_shri = len(sachin[(sachin["Runs"]>=50) & (sachin["Versus"]=="Sri Lanka")])
p_least_50_runs_given_shri = least_50_runs_given_shri / shri

morethan_30_balls = len(sachin[(sachin["Balls_faced"]>30)])
morethan_30_balls_morethan_30_runs = len(sachin[(sachin["Runs"]>30) & (sachin["Balls_faced"]>30)])
p_morethan_30_balls_morethan_30_runs = morethan_30_balls_morethan_30_runs / morethan_30_balls

runs_0 = len(sachin[sachin["Runs"]==0])
caught = len(sachin[(sachin["mode_of_dismissal"].str.startswith("c")) & (sachin["Runs"]==0)])
p_runs_0_caught = caught / runs_0

england = len(sachin[(sachin["Versus"]=="England")])
least_25_balls_eng = len(sachin[(sachin["Balls_faced"]>=25) & (sachin["Versus"]=="England")])
p_least_30_runs_given_eng = least_25_balls_eng / england

print(f"P(scoring more than 50 runs | playing against Sri Lanka):{p_least_50_runs_given_shri:.3f}")
print(f"P(scoring more than 30 runs | facing more than 30 balls):{p_morethan_30_balls_morethan_30_runs:.3f}")
print(f"P(scoring 0 runs | being dismissed by catch):{p_runs_0_caught:.3f}")
print(f"P(scoring more than 25 balls | playing against England):{p_least_30_runs_given_eng:.3f}")
```

```python
df = pd.concat([df_2015, df_2016, df_2017])
df
```

```python
df[["Hours","Minutes","Seconds"]] = df["Official Time"].str.split(":", expand=True).astype(int)
df["finish_time_in_seconds"] = df["Hours"]*3600 + df["Minutes"]*60 + df["Seconds"]
```

```python
df[["hours","minutes","seconds"]]=df["Pace"].str.split(":", expand=True).astype(int)
df["pace_in_seconds"] = df["hours"]*3600 + df["minutes"]*60 + df["seconds"]
```

```python
condition_A = df["finish_time_in_seconds"]<=3*3600
condition_B = df["pace_in_seconds"]<8*60
```

```python
#probability od A
p_A = len(df[condition_A])/len(df)
print(f"P(A):{p_A:.3f}")

#probability of B
p_B = len(df[condition_B])/len(df)
print(f"P(B):{p_B:.3f}")

#probability p(B|A)
p_B_given_A = len(df[condition_A & condition_B])/len(df[condition_A])
print(f"P(B|A):{p_B_given_A:.3f}")
```

```python
# Bayes theorem
p_A_given_B = (p_B_given_A * p_A) / p_B
print(f"P(A|B):{p_A_given_B:.3f}")
```

```python
import pandas as pd
df=pd.read_csv("loan_approval_dataset.csv")

import pandas as pd

# Load dataset

df.columns = df.columns.str.strip()

def bayes_probability(df):
    condition_A = df['loan_status'] == 1
    condition_B = df['education'].str.lower() == 'non_graduate'

    P_A = condition_A.mean()
    P_B = condition_B.mean()
    P_B_given_A = (condition_A & condition_B).sum() / condition_A.sum()
    P_A_given_B = (P_B_given_A * P_A) / P_B
    return P_A_given_B

# Initial probability
print("Initial P(Loan Approved | Non-Graduate):", bayes_probability(df))

# Function to add new record with 3 conditions
def add_new_customer(df, education, self_employed, loan_status):
    new_record = pd.DataFrame({
        'loan_id':[len(df)+1],
        'no_of_dependents':[0],
        'education':[education],
        'self_employed':[self_employed],
        'loan_status':[loan_status]
    })
    df = pd.concat([df, new_record], ignore_index=True)
    print(f"After adding {education}, Self Employed={self_employed}, Loan Status={loan_status}, new probability:", bayes_probability(df))
    return df

# Example usage:
df = add_new_customer(df, "Graduate", "No", 1)        # New Customer 1
df = add_new_customer(df, "Non_Graduate", "Yes", 0)   # New Customer 2
df = add_new_customer(df, "Non_Graduate", "No", 1)    # New Customer 3
```

```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv("loan_approval_dataset.csv")
df.columns = df.columns.str.strip()
print(df.head())

# 2. Initial counts
grad_total = len(df[df["education"] == "Graduate"])
grad_approved = len(df[(df["education"] == "Graduate") & (df["loan_status"] == "Approved")])

nongrad_total = len(df[df["education"] == "Not Graduate"])
nongrad_approved = len(df[(df["education"] == "Not Graduate") & (df["loan_status"] == "Approved")])

# 3. Initial probabilities (with safe division)
print("\nInitial probabilities")
if grad_total > 0:
    print("P(Approved | Graduate) =", round(grad_approved / grad_total, 4))
else:
    print("P(Approved | Graduate) = Not Defined (no Graduate records)")

if nongrad_total > 0:
    print("P(Approved | Not Graduate) =", round(nongrad_approved / nongrad_total, 4))
else:
    print("P(Approved | Not Graduate) = Not Defined (no Non-Graduate records)")

# -------------------------------
# 4. Add New Customer 1
education = input("\nCustomer 1 Education (Graduate/Not Graduate): ")
loan = input("Loan status (Approved/Rejected): ")

if education == "Graduate":
    grad_total += 1
    if loan == "Approved":
        grad_approved += 1
else:
    nongrad_total += 1
    if loan == "Approved":
        nongrad_approved += 1

print("\nUpdated probabilities after Customer 1")
if grad_total > 0:
    print("P(Approved | Graduate) =", round(grad_approved / grad_total, 4))
else:
    print("P(Approved | Graduate) = Not Defined")

if nongrad_total > 0:
    print("P(Approved | Not Graduate) =", round(nongrad_approved / nongrad_total, 4))
else:
    print("P(Approved | Not Graduate) = Not Defined")
```
