```python
#Line Plot
df = pd.read_csv("tips.csv")
avg_tip_by_day = df.groupby("day")["tip"].mean()
corr = df[["total_bill","tip","size"]].corr()

#functional syntax first
plt.figure(figsize=(12,8))


#Line
plt.subplot(2,3,1)
plt.plot(df.index,df["total_bill"])
plt.title("Line Plot")

#Bar
plt.subplot(2,3,2)
plt.bar(avg_tip_by_day.index,avg_tip_by_day.values)
plt.title("Bar Plot")

#Histogram
plt.subplot(2,3,3)
plt.hist(df["total_bill"],bins=10)
plt.title("Histogram Plot")

#Box
plt.subplot(2,3,4)
plt.boxplot(df["tip"])
plt.title("Box Plot")

#scatter
plt.subplot(2,3,5)
plt.scatter(df["total_bill"], df["tip"],marker="*")
plt.title("Scatter Plot")

#heatmap
plt.subplot(2,3,6)
plt.imshow(corr)
plt.title("Heatmap")
plt.colorbar()

plt.tight_layout()
plt.show()

```

```python
#oo style
fig, axes = plt.subplots(2, 3, figsize=(12,8))

#line
axes[0,0].plot(df.index, df["total_bill"])
axes[0,0].set_title("line plot")

#Bar
axes[0,1].bar(avg_tip_by_day.index,avg_tip_by_day.values)
axes[0,1].set_title("Bar plot")

#Histogram
axes[0,2].hist(df["total_bill"],bins=10)
axes[0,2].set_title("Histogram")

#Box
axes[1,0].boxplot(df["tip"])
axes[1,0].set_title("Box plot")

#scatter
axes[1,1].scatter(df["total_bill"],df["tip"])
axes[1,1].set_title("Scatter plot")

#Heatmap
img = axes[1,2].imshow(corr)
axes[1,2].set_title("Heatmap")
fig.colorbar(img, ax=axes[1,2])

plt.tight_layout()
plt.show()
```
