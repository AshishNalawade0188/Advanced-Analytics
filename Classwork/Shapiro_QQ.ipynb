```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
df = pd.read_csv("tips.csv")
total_bill_series = df["total_bill"]
total_bill_series
```

```python
shapiro_test = stats.shapiro(total_bill_series)
print(shapiro_test)
print(shapiro_test.statistic)
print(shapiro_test.pvalue)
alpha = 0.05
```

```python
if shapiro_test.pvalue > alpha:
    print("data is normally distributed")
else:
    print("data is not normally distributed")
```

```python
stats.probplot(total_bill_series,dist="norm",plot=plt)
plt.title("QQ plot of total bill")
plt.xlabel("Therotical quantile (Normal destribution)")
plt.ylabel("Ordered values(total bill)")
plt.grid(True)
plt.show()
```

```python
df = pd.read_csv("diabetes.csv")
bp = df["BloodPressure"]
bp_test = stats.shapiro(bp)
print(bp_test)
print(bp_test.statistic)
print(bp_test.pvalue)
alpha = 0.05

if bp_test.pvalue > alpha:
    print("Data is normally distributed")
else:
    print("data is not normally distributed")
    

stats.probplot(bp,dist="norm",plot=plt)
plt.title("Blood pressure")
plt.xlabel("Normal destribution")
plt.ylabel("pvalue")
plt.grid(True)
plt.show()
```
