```python
survival_rate = df.groupby("pclass")["survived"].mean()

print(survival_rate)

highest_class = survival_rate.idxmax()
highest_rate = survival_rate.max()

print(f"Passenger class {highest_class} had the highest survival rate: {highest_rate:.2f}")
```

```python
survival_rate = df.groupby("gender")["survived"].mean()

print(survival_rate)

highest_gender = survival_rate.idxmax()
highest_rate = survival_rate.max()

print(f"gender {highest_gender} had the highest survival rate: {highest_rate:.2f}")
```
