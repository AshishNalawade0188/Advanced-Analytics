```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_2015 = pd.read_csv("marathon_results_2015.csv")
df_2016 = pd.read_csv("marathon_results_2016.csv")
df_2017 = pd.read_csv("marathon_results_2017.csv")

df =pd.concat([df_2015, df_2016, df_2017], ignore_index=True)
Original_row_count = len(df)
Original_row_count
df.describe()
```

```python
sampled_df = df.sample(frac=0.1, random_state=42)
sampled_row_count = len(sampled_df)
sampled_row_count
sampled_df.describe()
```

```python
#systematic random sample

sample_df_systematic = df.iloc[::7, :]
sample_df_systematic_row_count = len(sample_df_systematic)
print(sample_df_systematic_row_count)
sample_df_systematic.head()
sample_df_systematic.describe()
```

```python
#stratified random sample
total_males = len(df[df["M/F"] == "M"])
total_females = len(df[df["M/F"] == "F"])

sample_size = int(0.1 * Original_row_count)
sample_males = int(0.1 * total_males)
sample_females = int(0.1 * total_females)

print(sample_size, sample_males, sample_females)
```

```python
male_sample = df[df["M/F"] == "M"].sample(n=sample_males, random_state=42)
female_sample = df[df["M/F"] == "F"].sample(n=sample_females, random_state=42)
print(len(male_sample), len(female_sample))
```

```python
combined_sample_MF = pd.concat([male_sample,female_sample], ignore_index=True)
combined_sample_MF_len = len(combined_sample_MF)
combined_sample_MF_len
```

```python
#clustered sampling
unique_countries = df["Country"].unique()
print(len(unique_countries),unique_countries)
```

```python
np.random.seed(101)
selected_countries = np.random.choice(unique_countries,size=len(unique_countries)//5, replace=False)
selected_countries
```

```python
sample_df = df[df["Country"].isin(selected_countries)]
sample_row_count = len(sample_df)
print(sample_row_count)
```
