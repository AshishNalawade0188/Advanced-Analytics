```python
import matplotlib.pyplot as plt
import seaborn as sns

commuter_times=[16,8,35,17,13,15,15,5,16,25,20,20,12,10]

plt.figure(figsize=(10,6))
sns.boxplot(data=commuter_times,orient='h')
plt.title('BOx Plot of Commuter Times')
plt.xlabel('Minutes')
plt.show()

# Display outlier points
import numpy as np
q1=np.percentile(commuter_times,25)
q3=np.percentile(commuter_times,75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

outliers=[]
for x in commuter_times:
    if x<lower_bound or x> upper_bound:
        outliers.append(x)

print("Q1 =",q1)
print("Q3 =",q3)
print("IQR =",iqr)
print("Lower Bound =",lower_bound)
print("Upper Bound =",upper_bound)
print("Outliers Points =",outliers)

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("D:\\D A\\billionaires.csv")

# Assuming 'finalworth' is a column in df
finalworth=df['finalWorth']

# Calculate mean, median, skewness, kurtosis
mean_finalworth = finalworth.mean()
median_finalworth = finalworth.median()
skewness_finalworth = finalworth.skew()
kurtosis_finalworth = finalworth.kurt()

# Plot
plt.figure(figsize=(8,6))
sns.histplot(finalworth, kde=True, color='skyblue', bins=30)

# Mean line
plt.axvline(mean_finalworth, color='red', linestyle='--', 
            label=f'Mean: {mean_finalworth:.2f}')

# Median line
plt.axvline(median_finalworth, color='black', linestyle='--', 
            label=f'Median: {median_finalworth:.2f}')

# Title with skewness & kurtosis
plt.title(f'Distribution of Final Worth\nSkewness: {skewness_finalworth:.2f}, Kurtosis: {kurtosis_finalworth:.2f}')

plt.xlabel('Final Worth')
plt.ylabel('Frequency')
plt.legend()  # comment this line if you get an error
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(finalworth)
plt.show()
# Conclusion:Many and extremely huge outliers: we need to find and remove them
```

```python
finalworth=df['finalWorth']
Q1=df['finalWorth'].quantile(0.25)
Q3=df['finalWorth'].quantile(0.75)
IQR=Q3-Q1

lower_fence=Q1-(3*IQR)# 1.5 times IQR would be too low here so we are using 3 times IQR
upper_fence=Q3+(3*IQR)

df['is_extreme_outlier']=(df['finalWorth']>upper_fence) |(df['finalWorth']<lower_fence)

print(df['is_extreme_outlier'])
num_extreme_outliers=df['is_extreme_outlier'].sum()
print(f"Number of extreme outliers identified:{num_extreme_outliers}")

#Remove extreme outliers
df_new=df[df['is_extreme_outlier']==False].copy()
finalworth=df_new['finalWorth']
mean_finalworth=finalworth.mean()
median_finalworth=finalworth.median()
skewness_finalworth=finalworth.skew()
print(skewness_finalworth)

plt.figure(figsize=(8,6))
sns.histplot(finalworth, kde=True, color='skyblue', bins=30)

# Mean line
plt.axvline(mean_finalworth, color='red', linestyle='--', 
            label=f'Mean: {mean_finalworth:.2f}')

# Median line
plt.axvline(median_finalworth, color='black', linestyle='--', 
            label=f'Median: {median_finalworth:.2f}')

# Title with skewness & kurtosis
plt.title(f'Distribution of Final Worth\nSkewness: {skewness_finalworth:.2f}, Kurtosis: {kurtosis_finalworth:.2f}')

plt.xlabel('Final Worth')
plt.ylabel('Frequency')
plt.legend()  # comment this line if you get an error
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(finalworth)
plt.show()
# Conclusion:Many and extremely huge outliers: we need to find and remove them
```

```python
# Corelation amd covariance analysis of two clo
#Visualization using scatter plot and heatmap
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns   # for heatmap

df = pd.read_csv("D:\\D A\\weight-height.csv")
print(df.head())

# Separate data based on gender
df_male = df[df['Gender'] == 'Male']
df_female = df[df['Gender'] == 'Female']

# Calculate Correlation (Pearson)
corr_overall = df[['Height','Weight']].corr()
corr_male = df_male[['Height','Weight']].corr()
corr_female = df_female[['Height','Weight']].corr()

# Calculate Spearman correlation
spearman_overall = df[['Height','Weight']].corr(method='spearman')
spearman_male = df_male[['Height','Weight']].corr(method='spearman')
spearman_female = df_female[['Height','Weight']].corr(method='spearman')

# Calculate Covariance
cov_overall = df[['Height','Weight']].cov()
cov_male = df_male[['Height','Weight']].cov()
cov_female = df_female[['Height','Weight']].cov()

print("Overall:\n", corr_overall, "\nSpearman:\n", spearman_overall)
print("Male:\n", corr_male, "\nSpearman:\n", spearman_male)
print("Female:\n", corr_female, "\nSpearman:\n", spearman_female)
print("Covariance:\n", cov_overall, "\n", cov_male, "\n", cov_female)

# Create scatter plots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Overall data plot
axes[0].scatter(df['Height'], df['Weight'], color='green', alpha=0.5)
axes[0].set_title(
    f'Overall P:{corr_overall.loc["Height","Weight"]:.2f} | S:{spearman_overall.loc["Height","Weight"]:.2f}'
)
axes[0].set_xlabel('Height')
axes[0].set_ylabel('Weight')
axes[0].grid(True)

# Male data plot
axes[1].scatter(df_male['Height'], df_male['Weight'], color='blue', alpha=0.5)
axes[1].set_title(
    f'Male P:{corr_male.loc["Height","Weight"]:.2f} | S:{spearman_male.loc["Height","Weight"]:.2f}'
)
axes[1].set_xlabel('Height')
axes[1].set_ylabel('Weight')
axes[1].grid(True)

# Female data plot
axes[2].scatter(df_female['Height'], df_female['Weight'], color='red', alpha=0.5)
axes[2].set_title(
    f'Female P:{corr_female.loc["Height","Weight"]:.2f} | S:{spearman_female.loc["Height","Weight"]:.2f}'
)
axes[2].set_xlabel('Height')
axes[2].set_ylabel('Weight')
axes[2].grid(True)

plt.show()

# Heatmap for correlation matrix
plt.figure(figsize=(6,4))
sns.heatmap(corr_overall, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap (Overall)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("D:\\D A\\Sachin_Tendulkar.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Keep only required columns
df = df[['Runs', 'Balls_faced', 'Versus', 'mode_of_dismissal']]
df.dropna(inplace=True)

# Normalize text columns
df['Versus'] = df['Versus'].str.strip().str.lower()
df['mode_of_dismissal'] = df['mode_of_dismissal'].str.strip().str.lower()

total = len(df)
print(f"Total matches: {total}")

# ---------------- Marginal Probabilities ----------------
p_zero_runs = len(df[df['Runs'] == 0]) / total
p_srilanka = len(df[df['Versus'] == 'sri lanka']) / total
p_50plus = len(df[df['Runs'] >= 50]) / total
p_30balls = len(df[df['Balls_faced'] > 30]) / total

print("\nMarginal Probabilities:")
print(f"P(Runs=0): {p_zero_runs:.3f} ({p_zero_runs*100:.2f}%)")
print(f"P(Versus=Sri Lanka): {p_srilanka:.3f} ({p_srilanka*100:.2f}%)")
print(f"P(Runs ≥ 50): {p_50plus:.3f} ({p_50plus*100:.2f}%)")
print(f"P(Balls_faced > 30): {p_30balls:.3f} ({p_30balls*100:.2f}%)")

# ---------------- Joint Probabilities ----------------
p_srilanka_50plus = len(df[(df['Versus'] == 'sri lanka') & (df['Runs'] >= 50)]) / total
p_zero_caught = len(df[(df['Runs'] == 0) & (df['mode_of_dismissal'].str.startswith('c'))]) / total
p_30balls_30runs = len(df[(df['Balls_faced'] > 30) & (df['Runs'] > 30)]) / total
p_pakistan_lbw = len(df[(df['Versus'] == 'pakistan') & (df['mode_of_dismissal'].str.startswith('lbw'))]) / total

print("\nJoint Probabilities:")
print(f"P(Sri Lanka ∩ Runs ≥ 50): {p_srilanka_50plus:.3f} ({p_srilanka_50plus*100:.2f}%)")
print(f"P(Runs=0 ∩ Dismissal=Caught): {p_zero_caught:.3f} ({p_zero_caught*100:.2f}%)")
print(f"P(Balls>30 ∩ Runs>30): {p_30balls_30runs:.3f} ({p_30balls_30runs*100:.2f}%)")
print(f"P(Pakistan ∩ Dismissal=LBW): {p_pakistan_lbw:.3f} ({p_pakistan_lbw*100:.2f}%)")

# ---------------- Conditional Probabilities ----------------
# (a) Given Versus = Sri Lanka, probability Runs ≥ 50
p_srilanka_cond = p_srilanka_50plus / p_srilanka if p_srilanka > 0 else 0

# (b) Given Balls_faced > 30, probability Runs > 30
p_30balls_cond = p_30balls_30runs / p_30balls if p_30balls > 0 else 0

# (c) Given Runs = 0, probability Dismissal = Caught
p_zero_cond = p_zero_caught / p_zero_runs if p_zero_runs > 0 else 0

# (d) Given Versus = England, probability Balls_faced ≥ 25
p_england = len(df[df['Versus'] == 'england']) / total
p_england_25balls = len(df[(df['Versus'] == 'england') & (df['Balls_faced'] >= 25)]) / total
p_england_cond = p_england_25balls / p_england if p_england > 0 else 0

# (e) Given Dismissal starts with 'c' (caught), probability Runs = 0
p_caught = len(df[df['mode_of_dismissal'].str.startswith('c')]) / total
p_caught_zero = len(df[(df['mode_of_dismissal'].str.startswith('c')) & (df['Runs'] == 0)]) / total
p_caught_cond = p_caught_zero / p_caught if p_caught > 0 else 0

# (f) Given Dismissal starts with 'lbw', probability Runs = 0
p_lbw = len(df[df['mode_of_dismissal'].str.startswith('lbw')]) / total
p_lbw_zero = len(df[(df['mode_of_dismissal'].str.startswith('lbw')) & (df['Runs'] == 0)]) / total
p_lbw_cond = p_lbw_zero / p_lbw if p_lbw > 0 else 0

print("\nConditional Probabilities:")
print(f"P(Runs ≥ 50 | Versus=Sri Lanka): {p_srilanka_cond:.3f} ({p_srilanka_cond*100:.2f}%)")
print(f"P(Runs > 30 | Balls_faced > 30): {p_30balls_cond:.3f} ({p_30balls_cond*100:.2f}%)")
print(f"P(Dismissal=Caught | Runs=0): {p_zero_cond:.3f} ({p_zero_cond*100:.2f}%)")
print(f"P(Balls_faced ≥ 25 | Versus=England): {p_england_cond:.3f} ({p_england_cond*100:.2f}%)")
print(f"P(Runs=0 | Dismissal starts with 'c'): {p_caught_cond:.3f} ({p_caught_cond*100:.2f}%)")
print(f"P(Runs=0 | Dismissal starts with 'lbw'): {p_lbw_cond:.3f} ({p_lbw_cond*100:.2f}%)")


```
