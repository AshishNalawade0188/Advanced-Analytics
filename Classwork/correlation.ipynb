```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("weight-height.csv")
df
```

```python
df_male = df[df["Gender"] == "Male"]
df_female = df[df["Gender"] == "Female"]
```

```python
corr_overall = df[["Height","Weight"]].corr()
corr_male = df_male[["Height","Weight"]].corr()
corr_female = df_female[["Height","Weight"]].corr()
```

```python
sp_corr_overall = df[["Height","Weight"]].corr(method="spearman")
sp_corr_male = df_male[["Height","Weight"]].corr(method="spearman")
sp_corr_female = df_female[["Height","Weight"]].corr(method="spearman")
```

```python
cov_overall = df[["Height","Weight"]].cov()
cov_male = df_male[["Height","Weight"]].cov()
cov_female = df_female[["Height","Weight"]].cov()
```

```python
print("Overall Correlation:\n", corr_overall, "\nOverall Spearman Correlation:\n", sp_corr_overall)
print("\nMale Correlation:\n", corr_male,"\nMale Spearman Correlation:\n", sp_corr_male)
print("\nFemale Correlation:\n", corr_female,"\nFemale Spearman Correlation:\n", sp_corr_female)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18,6))
axes[0].scatter(df["Height"], df["Weight"],color="green", alpha=0.5)
axes[0].set_title("Overall")
axes[0].set_xlabel("Height")    
axes[0].set_ylabel("Weight")

axes[1].scatter(df_male["Height"], df_male["Weight"],color="blue", alpha=0.5)
axes[1].set_title("Male")
axes[1].set_xlabel("Height")
axes[1].set_ylabel("Weight")

axes[2].scatter(df_female["Height"], df_female["Weight"],color="red", alpha=0.5)
axes[2].set_title("Female")
axes[2].set_xlabel("Height")
axes[2].set_ylabel("Weight")
```
