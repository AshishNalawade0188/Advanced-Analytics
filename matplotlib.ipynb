```python
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\ashis\OneDrive\Desktop\CDAC\Advanced Analytics\Advanced-Analytics\Classwork\tips.csv")
```

```python
plt.figure(figsize=(6,4))
plt.plot(df.index, df["tip"],marker = "o")
plt.title("Line Plot -Tip Across reacord")
plt.xlabel("Record Index")
plt.ylabel("Tip")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.lineplot(x=df.index, y =df["tip"])
plt.title("Line Plot -Tip Across reacord")
plt.xlabel("Record Index")
plt.ylabel("Tip")
plt.show()
```

```python
avg_bill_day = df.groupby("day")["total_bill"].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_bill_day.index, avg_bill_day.values)
plt.title("Bar Plot - Average total bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()
```
